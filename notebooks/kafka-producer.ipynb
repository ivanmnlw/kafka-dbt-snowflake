```python
import json
import uuid
import os
import json
import random
from dotenv import load_dotenv
from pathlib import Path
from kafka import KafkaProducer
from faker import Faker
from time import sleep
```

```python
dotenv_path = Path('/resources/.env')
load_dotenv(dotenv_path=dotenv_path)
```

```python
kafka_host = os.getenv('KAFKA_HOST')
kafka_topic = os.getenv('KAFKA_TOPIC_NAME')
kafka_topic_partition = os.getenv('KAFKA_TOPIC_NAME')+"-1"
```

```python
print(kafka_topic)
```

```python
producer = KafkaProducer(bootstrap_servers=f'{kafka_host}:9092')
```

```python
faker = Faker()
```

```python
print(round(random.uniform(1.5, 5.7), 2))
```

```python
def get_data():
    return [
            uuid.uuid4().__str__(), # trip_id
            faker.random_element(elements=('app', 'website', 'offline')), # booking_source
            faker.date_time().strftime("%Y-%m-%d %H:%M:%S"), # booking_timestamp
            faker.random_int(min=1, max=100), # destination_location
            faker.random_int(min=1, max=1000), # driver_id
            float(faker.latitude()), # drivers_current_lat
            float(faker.longitude()), # drivers_current_lon
            faker.random_int(min=1, max=100), # drivers_location
            faker.random_int(min=1, max=5), # drivers_review
            faker.random_int(min=1, max=10), # estimated_arrival_time_in_m
            faker.random_int(min=15000, max=100000), # estimated_fare
            faker.random_int(min=1, max=5), # number_of_passenger
            faker.date_time().strftime("%Y-%m-%d %H:%M:%S"), # order_completion_time
            float(faker.latitude()), # passenger_arrival_lat
            float(faker.longitude()), # passenger_arrival_lon
            float(faker.latitude()), # passenger_current_lat
            float(faker.longitude()), # passenger_current_lon
            faker.date_time().strftime("%Y-%m-%d %H:%M:%S"), # passenger_pick_up_time
            faker.random_int(min=1, max=100), # pickup_location
            faker.random_element(elements=('completed', 'accepted', 'cancelled')), # status
            faker.random_int(min=1, max=100), # taxi_id
            faker.random_element(elements=('morning', 'afternoon', 'night')), # time_of_day
            faker.random_element(elements=('light', 'heavy')), # traffic_condition
            round(random.uniform(1.5, 10.7), 2), # trip_distance
            round(random.uniform(1.5, 10.7), 2), # trip_duration
            faker.random_int(min=1, max=5), # trip_rating
            uuid.uuid4().__str__(), # user_id
            faker.random_element(elements=('rainy', 'sunny', 'cloudy')) # weather_condition
    ]
```

```python
for i in range(100):
    columns = ["trip_id", "booking_source", "booking_timestamp", "destination_location", "driver_id", "drivers_current_lat", 
               "drivers_current_lon", "drivers_location", "drivers_review", "estimated_arrival_time_in_m", "estimated_fare",
               "number_of_passenger", "order_completion_time", "passenger_arrival_lat", "passenger_arrival_lon", "passenger_current_lat",
               "passenger_current_lon", "passenger_pick_up_time", "pickup_location", "status", "taxi_id", "time_of_day", 
               "traffic_condition", "trip_distance", "trip_duration", "trip_rating", "user_id", "weather_condition"
              ]
    data_list = get_data()
    json_data = dict(zip(columns, data_list))
    _payload = json.dumps(json_data).encode("utf-8")
    response = producer.send(topic=kafka_topic, value=_payload)
    sleep(2)
    print(_payload)
```
